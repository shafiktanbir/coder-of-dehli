```python

# load the data
from pprint import pprint
import json
with open("data.json","r") as d:
    data=json.load(d)
pprint(data)

type(data)
```

```python
print("Users and Their Connections:\n")
for user in data["users"]:
    print(f"  user Id '{user['id']}' '{user['name']}'  friend with '{user['friends']}' Like pages '{user['liked_pages']}'  ")


print("pages taht user like\n")
for page in data["pages"]:
    print(f" user Id of this user '{page['id']}'   '{page['name']}' ")
```
